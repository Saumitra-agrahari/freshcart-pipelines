# Part C: ML Pipeline with Mage (Iris Classifier)

This notebook demonstrates and documents the Mage pipeline blocks:
- `load_iris` (data_loader)
- `prepare_data` (transformer - 80/20 train/test split)
- `train_model` (transformer - DecisionTreeClassifier with test-gate @test)
- `save_results` (data_exporter - exports to `iris_predictions.csv`)
- `evaluate_model` (transformer - confusion matrix & classification metrics)
- `train_knn` (transformer - parallel branch KNeighborsClassifier)

```python
from iris_project.data_loaders.load_iris import load_data, test_output as test_load
from iris_project.transformers.prepare_data import transform, test_output as test_prep
from iris_project.transformers.train_model import train, test_accuracy
from iris_project.data_exporters.save_results import export_data, test_export_file_created
from iris_project.transformers.evaluate_model import evaluate, test_evaluation_metrics
from iris_project.transformers.train_knn import train_knn, test_knn_accuracy

# 1. Load Data
df = load_data()
test_load(df)
print("load_iris: SUCCESS, shape=", df.shape)

# 2. Prepare Data (80/20 split)
split_df = transform(df)
test_prep(split_df)
print("prepare_data: SUCCESS, split counts=", split_df['split'].value_counts().to_dict())

# 3. Train Model (e.g. max_depth=3)
model_output = train(split_df, max_depth=3)
test_accuracy(model_output)
print("train_model: SUCCESS, accuracy=", model_output['accuracy'])

# 4. Save Results
exported_df = export_data(model_output)
test_export_file_created(exported_df)
print("save_results: SUCCESS, exported rows=", len(exported_df))

# 5. Evaluation Branch
eval_output = evaluate(model_output)
test_evaluation_metrics(eval_output)
print("evaluate_model: SUCCESS")

# 6. Parallel Branch: KNN
knn_output = train_knn(split_df)
test_knn_accuracy(knn_output)
print("train_knn: SUCCESS, accuracy=", knn_output['accuracy'])
```
